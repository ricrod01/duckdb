##### Valores nulos, distancias o montos negativos, fechas inconsistentes y cantidades de pasajeros sospechosas.

```python
calidad_yellow = con.execute(f"""
    SELECT
        COUNT(*) AS total_registros,
        COUNT(*) FILTER (WHERE passenger_count IS NULL) AS passenger_count_null,
        COUNT(*) FILTER (WHERE trip_distance IS NULL) AS trip_distance_null,
        COUNT(*) FILTER (WHERE trip_distance < 0) AS trip_distance_negativa,
        COUNT(*) FILTER (WHERE passenger_count < 0) AS passenger_count_negativo,
        COUNT(*) FILTER (WHERE tpep_dropoff_datetime < tpep_pickup_datetime) AS fechas_inconsistentes
    FROM read_parquet('{RUTA_YELLOW}')
""").df()

calidad_yellow
```
